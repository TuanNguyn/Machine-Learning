Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
y_pred = x @ w
mse = torch.mean(torch.square(y - y_pred))
mse.backward()
optimizer.step()
print("Learned coefficients:")
print(w)
```

Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
y2 = C - D - 1
y3 = A + C - 8
y4 = B - D - 2
sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
optimizer.zero_grad()
sqerr.backward()
optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau

```python
import torch

#Tạo Tensor X và reshape theo các kích thước yêu cầu
X = torch.arange(24)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)
print("X =", X)
print("X reshape (4, 6):\n", X_4x6)
print("X reshape (2, 12):\n", X_2x12)
print("X reshape (2, 3, 4):\n", X_2x3x4)

#Các phép toán trên hai Tensor kích thước 3 x 4
A = torch.arange(1, 13, dtype=torch.float32).reshape(3, 4)
B = torch.arange(2, 26, 2, dtype=torch.float32).reshape(3, 4)
print("\nA =\n", A)
print("B =\n", B)
print("A + B =\n", A + B)
print("A * B =\n", A * B)
print("A / B =\n", A / B)
print("Trung bình theo cột =", A.mean(dim=0))
print("Độ lệch chuẩn theo cột =", A.std(dim=0))

#Tính dy/dx bằng autograd và kiểm tra thủ công
x_autograd = torch.tensor(2.0, requires_grad=True)
y_autograd = 2 * x_autograd**2 + 5 * x_autograd + 3
dy_dx = torch.autograd.grad(y_autograd, x_autograd)[0]
dy_dx_manual = 4 * x_autograd + 5
print("\ny tại x = 2:", y_autograd.item())
print("dy/dx bằng autograd:", dy_dx.item())
print("dy/dx thủ công:", dy_dx_manual.item())
print("Hai kết quả bằng nhau:", torch.isclose(dy_dx, dy_dx_manual).item())

#Huấn luyện y = 2x^2 - 3x + 5 bằng Gradient Descent
x_train = torch.linspace(-5, 5, 100).reshape(-1, 1)
y_train = 2 * x_train**2 - 3 * x_train + 5
coefficients = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.Adam([coefficients], lr=0.05)
x_features = torch.cat([x_train**2, x_train, torch.ones_like(x_train)], dim=1)

for _ in range(2000):
    y_pred = x_features @ coefficients
    loss = torch.mean((y_pred - y_train)**2)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

learned = coefficients.detach().flatten()
print("\nHệ số tìm được [a, b, c] trong y = ax^2 + bx + c:", learned)
print("Sai số cuối cùng:", loss.item())
```
